# Pengolahan Citra Digital
## UTS - Menggunakan Gambar Sendiri

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img1 = cv2.imread('/content/orang.jpg')
img2 = cv2.imread('/content/barang.jpg')

img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

plt.imshow(img1); plt.title('Gambar 1'); plt.axis('off'); plt.show()
plt.imshow(img2); plt.title('Gambar 2'); plt.axis('off'); plt.show()
```

## Grayscale

```python
gray1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)

plt.imshow(gray1, cmap='gray'); plt.title('Gray 1'); plt.axis('off'); plt.show()
plt.imshow(gray2, cmap='gray'); plt.title('Gray 2'); plt.axis('off'); plt.show()
```

## Brightness & Contrast

```python
bright = cv2.convertScaleAbs(img1, beta=60)
contrast = cv2.convertScaleAbs(img1, alpha=1.8)

plt.imshow(bright); plt.title('Brightness'); plt.axis('off'); plt.show()
plt.imshow(contrast); plt.title('Contrast'); plt.axis('off'); plt.show()
```

```python
bright = cv2.convertScaleAbs(img2, beta=60)
contrast = cv2.convertScaleAbs(img2, alpha=1.8)

plt.imshow(bright); plt.title('Brightness'); plt.axis('off'); plt.show()
plt.imshow(contrast); plt.title('Contrast'); plt.axis('off'); plt.show()
```

## Transformasi Geometri

```python
rows, cols, _ = img1.shape
M = np.float32([[1,0,50],[0,1,50]])
trans = cv2.warpAffine(img1, M, (cols, rows))

M_rot = cv2.getRotationMatrix2D((cols/2, rows/2),45,1)
rot = cv2.warpAffine(img1, M_rot, (cols, rows))

scale = cv2.resize(img1, None, fx=0.5, fy=0.5)
flip = cv2.flip(img1,1)

plt.imshow(trans); plt.title('Translasi'); plt.axis('off'); plt.show()
plt.imshow(rot); plt.title('Rotasi'); plt.axis('off'); plt.show()
plt.imshow(scale); plt.title('Scaling'); plt.axis('off'); plt.show()
plt.imshow(flip); plt.title('Flip'); plt.axis('off'); plt.show()
```

```python
rows, cols, _ = img2.shape
M = np.float32([[1,0,50],[0,1,50]])
trans = cv2.warpAffine(img2, M, (cols, rows))

M_rot = cv2.getRotationMatrix2D((cols/2, rows/2),45,1)
rot = cv2.warpAffine(img2, M_rot, (cols, rows))

scale = cv2.resize(img2, None, fx=0.5, fy=0.5)
flip = cv2.flip(img2,1)

plt.imshow(trans); plt.title('Translasi'); plt.axis('off'); plt.show()
plt.imshow(rot); plt.title('Rotasi'); plt.axis('off'); plt.show()
plt.imshow(scale); plt.title('Scaling'); plt.axis('off'); plt.show()
plt.imshow(flip); plt.title('Flip'); plt.axis('off'); plt.show()
```

## Histogram Equalization

```python
eq = cv2.equalizeHist(gray1)
plt.imshow(eq, cmap='gray'); plt.title('Histogram Equalization'); plt.axis('off'); plt.show()
```

```python
eq = cv2.equalizeHist(gray2)
plt.imshow(eq, cmap='gray'); plt.title('Histogram Equalization'); plt.axis('off'); plt.show()
```

## Filtering

```python
blur = cv2.GaussianBlur(img1,(5,5),0)
kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
sharp = cv2.filter2D(img1,-1,kernel)

plt.imshow(blur); plt.title('Blur'); plt.axis('off'); plt.show()
plt.imshow(sharp); plt.title('Sharpen'); plt.axis('off'); plt.show()
```

```python
blur = cv2.GaussianBlur(img2,(5,5),0)
kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
sharp = cv2.filter2D(img2,-1,kernel)

plt.imshow(blur); plt.title('Blur'); plt.axis('off'); plt.show()
plt.imshow(sharp); plt.title('Sharpen'); plt.axis('off'); plt.show()
```
