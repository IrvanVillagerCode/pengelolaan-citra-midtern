```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load gambar
img1 = cv2.imread('/content/orang.jpg')
img2 = cv2.imread('/content/barang.jpg')

# Convert BGR ke RGB
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

plt.imshow(img1)
plt.title("Gambar 1")
plt.show()

plt.imshow(img2)
plt.title("Gambar 2")
plt.show()
```

```python
gray1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)

plt.imshow(gray1, cmap='gray')
plt.title("Grayscale 1")
plt.show()

plt.imshow(gray2, cmap='gray')
plt.title("Grayscale 2")
plt.show()
```

```python
# Brightness
bright = cv2.convertScaleAbs(img1, beta=50)

# Contrast
contrast = cv2.convertScaleAbs(img1, alpha=1.5)

plt.imshow(bright)
plt.title("Brightness")
plt.show()

plt.imshow(contrast)
plt.title("Contrast")
plt.show()
```

```python
# Brightness
bright = cv2.convertScaleAbs(img2, beta=50)

# Contrast
contrast = cv2.convertScaleAbs(img2, alpha=1.5)

plt.imshow(bright)
plt.title("Brightness")
plt.show()

plt.imshow(contrast)
plt.title("Contrast")
plt.show()
```

```python
# a. Translasi

rows, cols, _ = img1.shape
M = np.float32([[1, 0, 50], [0, 1, 50]])
trans = cv2.warpAffine(img1, M, (cols, rows))

plt.imshow(trans)
plt.title("Translasi")
plt.show()
```

```python
rows, cols, _ = img2.shape
M = np.float32([[1, 0, 50], [0, 1, 50]])
trans = cv2.warpAffine(img2, M, (cols, rows))

plt.imshow(trans)
plt.title("Translasi")
plt.show()
```

```python
# b. Rotasi

M = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rot = cv2.warpAffine(img1, M, (cols, rows))

plt.imshow(rot)
plt.title("Rotasi")
plt.show()
```

```python
M = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rot = cv2.warpAffine(img2, M, (cols, rows))

plt.imshow(rot)
plt.title("Rotasi")
plt.show()
```

```python
# c. Scaling
scale = cv2.resize(img1, None, fx=0.5, fy=0.5)

plt.imshow(scale)
plt.title("Scaling")
plt.show()
```

```python
scale = cv2.resize(img2, None, fx=0.5, fy=0.5)

plt.imshow(scale)
plt.title("Scaling")
plt.show()
```

```python
# d. Flipping

flip = cv2.flip(img1, 1)

plt.imshow(flip)
plt.title("Flip")
plt.show()
```

```python
flip = cv2.flip(img2, 1)

plt.imshow(flip)
plt.title("Flip")
plt.show()
```

```python
# 5. Histogram Equalization
eq = cv2.equalizeHist(gray1)

plt.imshow(eq, cmap='gray')
plt.title("Histogram Equalization")
plt.show()
```

```python
eq = cv2.equalizeHist(gray2)

plt.imshow(eq, cmap='gray')
plt.title("Histogram Equalization")
plt.show()
```

```python
# 6. Filtering Spasial
  # a. Smoothing (Blur)
blur = cv2.GaussianBlur(img1, (5,5), 0)

plt.imshow(blur)
plt.title("Smoothing")
plt.show()
```

```python
blur = cv2.GaussianBlur(img2, (5,5), 0)

plt.imshow(blur)
plt.title("Smoothing")
plt.show()
```

```python
# b. Sharpening
kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
sharp = cv2.filter2D(img1, -1, kernel)

plt.imshow(sharp)
plt.title("Sharpening")
plt.show()
```

```python
kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
sharp = cv2.filter2D(img2, -1, kernel)

plt.imshow(sharp)
plt.title("Sharpening")
plt.show()
```
